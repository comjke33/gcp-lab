### 6. Task 5. Create a Pub/Sub Topic

Cloud Pub/Sub을 사용해서 콘솔에서 클라우드 리소스를 바로 만들 수 있다.

응용 프로그램이나 서비스에서 발생하는 이벤트 데이터를 다룰 때 이용한다.

Cloud Pub/Sub API는 기본적으로 enable되어있다.

그리고 Admin Role도 주어져있기 때문에 바로 topic을 생성할 수 있다.

  
All products > Analytics > Pub/Sub 으로 이동

  
Create topic 클릭

  
Topic ID에 'my-topic'으로 이름 짓고, Create 클릭

  
Topic name을 보면, 나의 Project ID가 포함되어 지어진 것을 확인할 수 있음.
